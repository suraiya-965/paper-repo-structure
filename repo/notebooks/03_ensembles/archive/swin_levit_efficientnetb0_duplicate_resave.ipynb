## Cell 6 — Config

```python
IMG_SIZE = (224, 224)
BATCH_SIZE = 32
filepaths = test_df['filepath'].values
val_filepaths = val_df['filepath'].values
```

## Cell 7 — Build the loading function(s)
LeViT-128 takes raw `float32` pixels in [0, 255] (its normalization is baked into the model). EfficientNetB0 needs `preprocess_input` applied explicitly, so it gets its own dataset.

```python
def load_float32(filepath):
    img = tf.io.read_file(filepath)
    img = tf.image.decode_image(img, channels=3, expand_animations=False)
    img = tf.image.resize(img, IMG_SIZE)
    return img  # stays float32, values 0-255

from tensorflow.keras.applications.efficientnet import preprocess_input as efficientnet_preprocess_input

def load_efficientnet(filepath):
    img = tf.io.read_file(filepath)
    img = tf.image.decode_image(img, channels=3, expand_animations=False)
    img = tf.image.resize(img, IMG_SIZE)
    img = efficientnet_preprocess_input(img)
    return img
```

## Cell 8 — Build the test/val dataset(s)

```python
def make_ds(paths, load_fn):
    ds = tf.data.Dataset.from_tensor_slices(paths)
    ds = ds.map(load_fn, num_parallel_calls=tf.data.AUTOTUNE)
    ds = ds.batch(BATCH_SIZE).prefetch(tf.data.AUTOTUNE)
    return ds

test_ds_float32 = make_ds(filepaths, load_float32)                # for LeViT-128
test_ds_efficientnet = make_ds(filepaths, load_efficientnet)      # for EfficientNetB0
val_ds_float32 = make_ds(val_filepaths, load_float32)
val_ds_efficientnet = make_ds(val_filepaths, load_efficientnet)

print("Datasets ready")
```

## Cell 9 — Load LeViT-128 and EfficientNetB0

```python
# --- LeViT-128: rebuild architecture exactly as in training, then load trained weights ---
levit_backbone = levit.LeViT128(input_shape=(224, 224, 3), num_classes=0, pretrained="imagenet")
levit_backbone.trainable = False

levit_inputs = layers.Input(shape=(224, 224, 3), dtype='float32')

class TorchNormalize(layers.Layer):
    def __init__(self, **kwargs):
        super().__init__(**kwargs)
        self.mean = tf.constant([0.485, 0.456, 0.406]) * 255.0
        self.std = tf.constant([0.229, 0.224, 0.225]) * 255.0

    def call(self, x):
        return (x - self.mean) / self.std

x = TorchNormalize()(levit_inputs)

x = levit_backbone(x)
x = layers.GlobalAveragePooling2D()(x)
x = layers.Dense(256, activation='relu')(x)
x = layers.Dropout(0.4)(x)
levit_outputs = layers.Dense(num_classes, activation='softmax')(x)

levit_model = models.Model(levit_inputs, levit_outputs)
levit_model.load_weights('/content/drive/MyDrive/thesis_levit_outputs/best_levit_model.keras')
print("LeViT-128 model rebuilt and weights loaded!")
```

```python
# --- EfficientNetB0: rebuild architecture, then load weights extracted from the .keras file ---
efficientnet_model_path = '/content/drive/MyDrive/best_efficientnet_model.keras'
print("Model file exists:", os.path.exists(efficientnet_model_path))

efficientnet_weights_extract_dir = '/content/efficientnet_weights_extracted'
if not os.path.exists(efficientnet_weights_extract_dir):
    with zipfile.ZipFile(efficientnet_model_path, 'r') as zip_ref:
        zip_ref.extractall(efficientnet_weights_extract_dir)

efficientnet_weights_path = None
for root, dirs, files in os.walk(efficientnet_weights_extract_dir):
    for fname in files:
        if fname.endswith('.weights.h5') or fname == 'model.weights.h5':
            efficientnet_weights_path = os.path.join(root, fname)
            break
print("Weights file found:", efficientnet_weights_path)

# preprocess_input is applied externally in load_efficientnet() below, so it's
# NOT baked into the model here (unlike DenseNet121, which bakes it in).
efficientnet_base = tf.keras.applications.EfficientNetB0(
    include_top=False,
    weights=None,
    input_shape=(224, 224, 3)
)

efficientnet_input = layers.Input(shape=(224, 224, 3))
x = efficientnet_base(efficientnet_input, training=False)
x = layers.GlobalAveragePooling2D()(x)
x = layers.Dense(256, activation='relu')(x)
x = layers.Dropout(0.4)(x)
efficientnet_output = layers.Dense(num_classes, activation='softmax')(x)

efficientnet_model = models.Model(efficientnet_input, efficientnet_output)
efficientnet_model.load_weights(efficientnet_weights_path)
print("EfficientNetB0 rebuilt and weights loaded successfully")
```

## Cell 10 — Predict on test + val sets, sanity-check accuracy

```python
pred_levit = levit_model.predict(test_ds_float32, verbose=1)
acc_levit = accuracy_score(true_labels, np.argmax(pred_levit, axis=1))
print(f"LeViT-128 TEST accuracy: {acc_levit:.4f}")

pred_efficientnet = efficientnet_model.predict(test_ds_efficientnet, verbose=1)
acc_efficientnet = accuracy_score(true_labels, np.argmax(pred_efficientnet, axis=1))
print(f"EfficientNetB0 TEST accuracy: {acc_efficientnet:.4f}")

val_pred_levit = levit_model.predict(val_ds_float32, verbose=1)
val_pred_efficientnet = efficientnet_model.predict(val_ds_efficientnet, verbose=1)
print("Val predictions ready")
```

## Cell 11 — Ensemble (simple average) + accuracy comparison

```python
ensemble_probs = (pred_swin + pred_levit + pred_efficientnet) / 3.0
ensemble_preds = np.argmax(ensemble_probs, axis=1)

acc_ensemble = accuracy_score(true_labels, ensemble_preds)

print("=" * 55)
print("RESULTS (Simple Average Ensemble)")
print("=" * 55)
print(f"Swin Tiny accuracy:          {acc_swin:.4f}")
print(f"LeViT-128 accuracy:          {acc_levit:.4f}")
print(f"EfficientNetB0 accuracy:          {acc_efficientnet:.4f}")
print(f"Ensemble (avg) accuracy:     {acc_ensemble:.4f}")
```

## Cell 12 — Weighted ensemble (weights tuned on VAL set, evaluated once on TEST set)

```python
best_w = (1/3, 1/3, 1/3)
best_val_acc = accuracy_score(
    val_true_labels,
    np.argmax((val_pred_swin + val_pred_levit + val_pred_efficientnet) / 3.0, axis=1)
)

step = 0.05
for w_swin in np.arange(0.0, 1.0 + step, step):
    for w_levit in np.arange(0.0, 1.0 - w_swin + step, step):
        w_efficientnet = 1.0 - w_swin - w_levit
        if w_efficientnet < -1e-9:
            continue
        w_efficientnet = max(w_efficientnet, 0.0)
        val_probs = (w_swin * val_pred_swin) + (w_levit * val_pred_levit) + (w_efficientnet * val_pred_efficientnet)
        val_acc = accuracy_score(val_true_labels, np.argmax(val_probs, axis=1))
        if val_acc > best_val_acc:
            best_val_acc = val_acc
            best_w = (round(w_swin, 2), round(w_levit, 2), round(w_efficientnet, 2))

print(f"Best weighted combo (found on VAL set) [w_swin, w_levit, w_efficientnet]: {best_w}")
print(f"Val accuracy with this combo: {best_val_acc:.4f}")

w_swin_final, w_levit_final, w_efficientnet_final = best_w
test_probs_weighted = (w_swin_final * pred_swin) + (w_levit_final * pred_levit) + (w_efficientnet_final * pred_efficientnet)
ensemble_preds_weighted = np.argmax(test_probs_weighted, axis=1)
acc_ensemble_weighted = accuracy_score(true_labels, ensemble_preds_weighted)

print(f"\nFinal TEST accuracy with tuned weights: {acc_ensemble_weighted:.4f}")
```

## Cell 13 — Classification report

```python
report = classification_report(true_labels, ensemble_preds, target_names=class_names)
print(report)
```
